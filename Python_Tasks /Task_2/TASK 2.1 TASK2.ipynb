1. Find the Shortest Word in a Sentence

```python
s="Python makes coding Easy"
word=s.split()
min=word[0]
for ele in word:
  if len(ele)<len(min):
    min=ele
print(min)
```

4 . Find the First Longest Word When Ties Occur

```python
s = "Python makes coding useful"
words = s.split()
longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word

print(longest)
```

6. Count Vowels in a String

```python
s = "Education"
count = 0

for ch in s:
    if ch.lower() in "aeiou":
        count += 1

print(count)
```

7. Count Consonants in a String

```python
s = "Python 123!"
count = 0
for ch in s:
    if ch.isalpha() and ch.lower() not in "aeiou":
        count += 1
print(count)
```

8. Find the Most Frequent Character

```python
s="banana"
s=s.replace(" ","")
max_count=0
max_char=""
for ch in s.lower():
  count=s.count(ch)
  if count>max_count:
    max_count=count
    max_char=ch
print(max_char)
```

9. Find the Least Frequent Character

15. Create a Dictionary of nu,bers and their Even/Odd Status

```python
numbers = [1, 2, 3, 4, 5]
result = {}
for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```

16. Find the most Frequent Word in Sentence

```python
s = "Python is easy and Python is powerful"
words = s.lower().split()
max_count = 0
most_frequent = ""
for word in words:
    count = words.count(word)
    if count > max_count:
        max_count = count
        most_frequent = word
print(most_frequent)
```

20. Find the longest word that appears only Once

```python
s = "Python makes coding useful Python"

words = s.lower().split()

longest = ""

for word in words:
    if words.count(word) == 1 and len(word) > len(longest):
        longest = word

print(longest)
```
