# Q1: Find the Frequency of Each Word in a Sentence

## Pseudocode

1. Take a sentence as input.
2. Split the sentence into words.
3. Create an empty dictionary.
4. Traverse each word.
5. If the word already exists, increase its count.
6. Otherwise, add the word with count 1.
7. Return the dictionary.

```python
# ------1------
def word_frequency(text):
    words=text.split()
    result={}
    for word in words:
        if word in result:
            result[word]+=1
        else:
            result[word]=1
    return result
```

```python
print(word_frequency("python is easy and python is powerful"))
```

 # Q2 Character Frequency Count
1. Take a word as input.
2. Create an empty dictionary called frequency.
3. For each character in the word:
      a. If the character already exists in the dictionary:
             Increase its count by 1.
      b. Otherwise:
             Add the character to the dictionary with count 1.
4. Return the frequency dictionary.
STOP

```python
# ------2------
def char_frequency(text):
    result={}
    for ch in text:
        if ch in result:
            result[ch]+=1
        else:
            result[ch]=1
    return result
```

# Q3. Find the First Non-Repeating Character in a Word

Pseudocode:

START
1. Take a word as input.
2. Create an empty dictionary called frequency.
3. Traverse the word and count the occurrences of each character.
4. Traverse the word again from left to right.
5. For each character:
      a. If its frequency is 1:
             Return that character immediately.
6. If no character has frequency 1:
      Return None.
STOP

```python
# ------3------
def first_non_rep_char(word):
    for ch in word:
        if word.count(ch)==1:
            return ch
    return None
```

```python
print("Q3:", first_non_rep_char("swiss"))
```

# Q4. Check Whether Two Strings Are Anagrams

Pseudocode:

START
1. Take two strings as input.
2. Sort the characters of the first string.
3. Sort the characters of the second string.
4. Compare the two sorted results.
5. If both results are equal:
      Return True.
6. Otherwise:
      Return False.
STOP

```python
# ------4------
def str_anagram(s1,s2):
    if len(s1)!=len(s2):
        return False
    return sorted(s1)==sorted(s2)
```

```python
print("Q4:", str_anagram("listen", "silent"))
```

# Q5. Find the First Repeating Character in a String

Pseudocode:

START
1. Take a string as input.
2. Create an empty set called seen.
3. Traverse each character from left to right.
4. For each character:
      a. If the character already exists in seen:
             Return that character.
      b. Otherwise:
             Add the character to seen.
5. If no character repeats:
      Return None.
STOP

```python
# ------5------
def first_rep_char(text):
    for ch in text:
        if text.count(ch)>1:
            return ch
    return None
```

# Q6. Move All Zeros to the End of a List

Pseudocode:

START
1. Take a list of numbers as input.
2. Create an empty list called result.
3. Traverse each number in the original list.
4. If the number is not zero:
      Add it to result.
5. Calculate the number of zeros in the original list.
6. Add all those zeros to the end of result.
7. Return result.
STOP

```python
# ------6------
def move_zeros_to_end(nums):
    result=[]
    for i in nums:
        if i!=0:
            result.append(i)
    zeros_count=len(nums)-len(result)
    result.extend([0]*zeros_count)
    return result
```

```python
print("Q6:", move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))
```

# Q7. Find the Second Highest Distinct Number in a List

Pseudocode:

START
1. Take a list of numbers as input.
2. Remove duplicate numbers.
3. Sort the distinct numbers in ascending order.
4. Check the number of distinct elements.
5. If fewer than two distinct elements exist:
      Return None.
6. Otherwise:
      Return the second-last element of the sorted list.
STOP

```python
# ------7------
def second_high_distinct_num(nums):
    result=sorted(set(nums))
    if len(result)<2:
        return None
    return result[-2]
```

```python
print("Q7:", second_high_distinct_num([10, 20, 40, 30, 40, 20]))
```

# Q8. Remove Duplicates Without Changing the Original Order

Pseudocode:

START
1. Take a list as input.
2. Create an empty list called result.
3. Traverse each element in the original list.
4. For each element:
      a. If the element is not already in result:
             Add it to result.
      b. Otherwise:
             Skip the element.
5. Return result.
STOP

```python
# ------8------
def remove_dup(nums):
    result=[]
    for num in nums:
        if num not in result:
            result.append(num)
    return result
```

```python
print("Q8:", remove_dup([4, 2, 4, 5, 2, 7, 5]))
```

# Q9. Find the Missing Number from 1 to N

Pseudocode:

START
1. Take a list of numbers and the value N as input.
2. Calculate the expected sum of numbers from 1 to N.
      Expected sum = N × (N + 1) / 2
3. Calculate the sum of the given list.
4. Subtract the actual sum from the expected sum.
5. Return the difference as the missing number.
STOP

```python
# ------9------
def missing_num(nums,n):
    sum_n=n*(n+1)//2
    s1=sum(nums)
    return sum_n-s1
```

```python
print("Q9:", missing_num([1, 2, 4, 5, 6], 6))
```

# Q10. Find Two Numbers Whose Sum Equals a Target

Pseudocode:

START
1. Take a list of numbers and a target value as input.
2. Create an empty set called seen.
3. Traverse each number from left to right.
4. For each number:
      a. Calculate the required number:
             Required = Target - Current number.
      b. If the required number exists in seen:
             Return [Required, Current number].
      c. Otherwise:
             Add the current number to seen.
5. If no valid pair is found:
      Return an empty list.
STOP

```python
# ------10------
def sum_equal_target(nums,target):
    seen=set()
    for num in nums:
        required=target-num
        if required in seen:
           return [required,num]
        seen.add(num)
    return []
```

```python
print("Q10:", sum_equal_target([2, 7, 11, 15], 9))
```

# Q11. Find the Number Closest to Zero

Pseudocode:

START
1. Take a list of numbers as input.
2. Compare the absolute values of the numbers.
3. Find the number with the smallest absolute value.
4. If two numbers are equally close to zero:
      Choose the positive number.
5. Return the selected number.
STOP

```python
# ------11------
def num_closeto_zero(nums):
    short=float('inf')
    for num in nums:
        shortest=abs(num)
        if shortest<short:
            short=shortest
    return short
```

```python
print("Q11:", num_closeto_zero([-8, -3, 2, 5, -2]))
```

# Q12. Find Common Elements Between Two Lists

Pseudocode:

START
1. Take two lists as input.
2. Create an empty list called result.
3. Traverse each element in the first list.
4. For each element:
      a. Check whether it exists in the second list.
      b. Check whether it is not already in result.
      c. If both conditions are true:
             Add it to result.
5. Return result.
STOP

```python
# ------12------
def common_ele_2list(l1,l2):
    result=[]
    for num in l1:
        if num in l2 and num not in result:
            result.append(num)
    return result
```

```python
print("Q12:", common_ele_2list([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
```

# Q13. Rotate a List to the Right by K Positions

Pseudocode:

START
1. Take a list and the number of rotations K as input.
2. If the list is empty:
      Return an empty list.
3. Calculate K modulo the length of the list.
4. If K equals zero:
      Return a copy of the original list.
5. Separate the last K elements from the list.
6. Separate the remaining elements.
7. Place the last K elements before the remaining elements.
8. Return the rotated list.
STOP

```python
# ------13------
def rotate_k_pos(nums,k):
    if not nums:
        return []
    k=k%len(nums)

    return nums[-k:]+nums[:-k] if k else nums[:]
```

```python
print("Q13:", rotate_k_pos([1, 2, 3, 4, 5], 2))
```

# Q14. Find the Maximum Sum of Two Consecutive Elements

Pseudocode:

START
1. Take a list containing at least two numbers.
2. Calculate the sum of the first two consecutive elements.
3. Store this sum as max_sum.
4. Traverse the remaining consecutive pairs.
5. For each pair:
      a. Calculate the sum of the two elements.
      b. If the current sum is greater than max_sum:
             Update max_sum.
6. Return max_sum.
STOP

```python
# ------14------
def max_sum_2consec(nums):
    max_sum=nums[0]+nums[1]
    for i in range(1,len(nums)-1):
        curr_sum=nums[i]+nums[i+1]
        if curr_sum>max_sum:
            max_sum=curr_sum
    return max_sum
```

```python
print("Q14:", max_sum_2consec([4, 7, 2, 9, 5]))
```

# Q15. Find All Unique Pairs Whose Sum Equals a Target

Pseudocode:

START
1. Take a list of numbers and a target value as input.
2. Create an empty set called seen.
3. Create an empty set called pairs.
4. Traverse each number in the list.
5. For each number:
      a. Calculate the required number:
             Required = Target - Current number.
      b. If the required number exists in seen:
             i. Create a pair containing the current number
                and the required number.
             ii. Arrange the two numbers in ascending order.
             iii. Add the pair to pairs.
      c. Add the current number to seen.
6. Sort all unique pairs by their first element.
7. Convert each pair into a list.
8. Return the list of pairs.
STOP

```python
# ------15------
def unique_pairs_sum_equal_target(nums,target):
    result=set()
    pairs=set()
    for num in nums:
        required=target-num
        if required in result:
            pair=tuple(sorted((num,required)))
            pairs.add(pair)
        result.add(num)
    return [list(pair) for pair in sorted(pairs)]
```

```python
print("Q15:", unique_pairs_sum_equal_target([2, 4, 3, 5, 7, 8, 1], 9))
```
